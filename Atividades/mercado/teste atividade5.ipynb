```python
import numpy as np

import pandas as pd
```

```python
pd.set_option("display.width", 200)
df = pd.read_csv("vendas_mercados.csv")
```

```python
df["Lucro"] = (df["Preco_Unitario"] - df["Custo_Unitario"]) * df["Quantidade"]
df["Faturamento"] = df["Preco_Unitario"] * df["Quantidade"]
```

```python
lucro = df.groupby("Mercado")["Lucro"].sum().sort_values(ascending=False)
print("\n--- Q1: LUCRO TOTAL POR MERCADO ---")
print(lucro.round(2))
print("Maior lucro:", lucro.idxmax(), "->", round(lucro.max(),2))
```

```python
pag = df.groupby(["Regiao", "Forma de Pagamento"]). size().unstack(fill_value=0)
print("\n--- Q2: VENDAS POR REGIÃO E FORMA DE PAGAMENTO ---")
print(pag)
print(pag.apply(lambda l: ", ".join(l[l == l.max()].index) + f" ({l.max()})", axis=1))
```
