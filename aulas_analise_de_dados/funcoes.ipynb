```python
import numpy as np
renda = np.array([5000, 4500, 6000, 4000])
gastos = np.array([4000, 3500, 6000, 3900])

lucro =  renda - gastos

print(f"Lucro dos meses: {lucro}")
```

```python
notas_alunos = np.array([
    10.0, 6.0, 9.6, 7.4, 8.8
])
notas_aprovados = notas_alunos[notas_alunos > 7]
print(f"Notas aprovados: {notas_aprovados}")

notas_reprovados = notas_alunos[~(notas_alunos >= 7)]
print(f"Notas reprovados: {notas_reprovados}")
print(notas_aprovados)
```

```python
vendas = np.array([
    [15000, 20000, 30000],
    [25000, 30000, 40000],
    [35000, 40000, 50000],
    [45000, 50000, 60000],
])

total_vendas_ano = np.sum(vendas)
print(total_vendas_ano)

total_por_trimetre = np.sum(vendas, axis=1)
print(total_por_trimetre)

comeco_de_trimetre = np.sum(vendas, axis=0)
print(comeco_de_trimetre)

maior_venda_ano = np.max(vendas)
print(f"maior_venda_ano: {maior_venda_ano}")

menor_venda_ano = np.min(vendas)
print(f"menor_venda_ano: {menor_venda_ano}")


```

```python
valor_produtos = np.array([
    10,15,20,30
])
#Recebe uma condição,true ou false
valor_promocao = np.where (valor_produtos >= 20, valor_produtos * 0.9, valor_produtos)

valor_promocao = np.round(valor_promocao, 2)

print(valor_promocao)
```

```python
import pandas as pd

df_produtos = pd.DataFrame({
    "Produto": ['Notebook', 'Celular', 'Monitor'],
    "Estoque": [10, 20, 5],
    "Valor": [2000, 1500, 500],
    "Cliente": ['Diego', 'Ana', 'Raimundo']
})
df_produtos["status"] = np.where(
    df_produtos["Estoque"] > 0,
    "Disponivel",
    "Esgotado"
)
print(df_produtos)
```

```python
agrupamento_count = df_produtos.groupby(["Cliente","Estoque"])["Valor"].count()

print(agrupamento_count)
```
