```python
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

df = pd.read_csv("alunos_senai_sig.csv")
df["Curso"] = df["Curso"].str.strip()
```

```python
resumo = df.groupby("Curso")["Media_Final"].agg(["max", "min"])
ax = resumo.plot(kind="bar", figsize=(9, 5), color=["#2e7d32", "#c62828"], width=0.7)
for c in ax.containers:
    ax.bar_label(c, fmt="%.1f", padding=2)
ax.legend(["Maior média", "Menor média"])
ax.set_title("Maiores e menores médias por curso")
ax.set_xlabel("Curso")
ax.set_ylabel("Média final")
ax.set_ylim(0, 11)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("grafico1_maior_menor_media.png", dpi=150)
plt.close()
```

```python
plt.figure(figsize=(11, 5.5))
cores = {"Design": "#1565c0", "Java": "#ef6c00"}
for curso in ["Design", "Java"]:
    notas = df[df["Curso"] == curso]["Media_Final"].reset_index(drop=True)
    x = range(1, len(notas) + 1)
    plt.plot(x, notas, marker="o", markersize=4, label=curso, color=cores[curso])
    pos = notas.idxmax()                       # posição da maior nota
    plt.scatter(pos + 1, notas[pos], s=160, color="red", marker="*", zorder=5)
    plt.annotate(f"maior nota ({notas[pos]:.1f})", (pos + 1, notas[pos]),
                 textcoords="offset points", xytext=(0, 12), ha="center",
                 color="red", fontweight="bold")
plt.title("Notas dos alunos: Design x Java")
plt.xlabel("Aluno (ordem no arquivo)")
plt.ylabel("Média final")
plt.ylim(0, 11)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig("grafico2_design_java.png", dpi=150)
plt.close()
```

```python
py = df[df["Curso"] == "Python"]
aprovados = (py["Media_Final"] >= 7).sum()
reprovados = (py["Media_Final"] < 7).sum()
plt.figure(figsize=(6, 6))
plt.pie([aprovados, reprovados],
        labels=[f"Aprovados ({aprovados})", f"Reprovados ({reprovados})"],
        autopct="%1.1f%%", startangle=90, colors=["#2e7d32", "#c62828"],
        textprops={"fontsize": 12})
plt.title("Python: aprovação (média mínima 7)")
plt.tight_layout()
plt.savefig("grafico3_pizza_python.png", dpi=150)
plt.close()

print(resumo)
print("Python -> aprovados:", aprovados, "| reprovados:", reprovados)
```
