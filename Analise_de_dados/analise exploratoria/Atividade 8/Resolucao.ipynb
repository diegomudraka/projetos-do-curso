```python
import pandas as pd

df = pd.read_csv("customers-10000.csv.crdownload")
```

```python
import matplotlib.pyplot as plt

# Como o dataset não indica explicitamente uma coluna de 'Status', assumimos as linhas como assinaturas registradas.
# 1. Encontrar o país e a cidade com maior número de assinaturas
most_common_country = df['Country'].value_counts().idxmax()
most_common_country_count = df['Country'].value_counts().max()

most_common_city = df['City'].value_counts().idxmax()
most_common_city_count = df['City'].value_counts().max()

print(f"País com maior número de assinaturas: {most_common_country} ({most_common_country_count} clientes)")
print(f"Cidade com maior número de assinaturas: {most_common_city} ({most_common_city_count} clientes)")

# 2. Gráfico de barras horizontais do Top 10 Países
top_10_countries = df['Country'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_10_countries.plot(kind='barh', color='skyblue').invert_yaxis()
plt.title('Top 10 Países com Mais Clientes')
plt.xlabel('Número de Clientes')
plt.ylabel('País')
plt.tight_layout()
plt.show()
```

```python
# 1. Converter a coluna Subscription Date para datetime
df['Subscription Date'] = pd.to_datetime(df['Subscription Date'])

# 2. Descobrir anos e meses com maior volume de vendas
df['Year_Month'] = df['Subscription Date'].dt.to_period('M')
sales_by_month = df['Year_Month'].value_counts().sort_index()

top_months = df['Year_Month'].value_counts().head(5)
print("Top 5 Meses/Anos com maior volume de vendas:")
print(top_months)

# 3. Gráfico de linha mostrando a evolução ao longo do tempo
plt.figure(figsize=(12, 6))
sales_by_month.plot(kind='line', marker='o', color='green', linewidth=2)
plt.title('Evolução do Volume de Assinaturas ao Longo do Tempo')
plt.xlabel('Período (Ano-Mês)')
plt.ylabel('Quantidade de Assinaturas')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
